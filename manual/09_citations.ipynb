# 9: Managing bibliography and references

ALPaca keeps track of every reference used during the computation, and can produce a bibliographic lists, in `.bib` format.

```python
import alpaca
from alpaca.biblio import citations, citations_context
import numpy as np
```

```python
a1 = alpaca.ALPcouplings({'cqL': 1.0}, basis='derivative_above', scale=4*np.pi*1000).match_run(5.2, 'VA_below')
```

```python
citations.generate_bibtex('09_results/biblio.running.bib')
```

```python
with citations_context(merge=True):
    br = alpaca.branching_ratio('a -> eta pi pi', 1.7, a1, 1000, integrator='no_rge')
    citations.generate_bibtex('09_results/biblio.branching_ratio.bib')
```

In statistical analyses, we can also generate a report containing only the experimental inputs:

```python
ma = np.logspace(-2, np.log10(4), 20)
fa = np.logspace(3, 8, 20)
x_ma, y_fa = np.meshgrid(ma, fa)
couplings = [alpaca.ALPcouplings({'cuR': np.diag([0,0,1])},
    4*np.pi*f, 'derivative_above').match_run(4, 'VA_below') for f in fa]
x_ma, y_couplings = np.meshgrid(ma, couplings)
```

```python
transitions_BKll = [
    'B+ -> K+ mu mu',
    'B+ -> K+ e e',
    'B0 -> K*0 mu mu',
    'B0 -> K*0 e e'
]

chi2_BKll = alpaca.statistics.get_chi2(
    transitions_BKll,
    x_ma,
    y_couplings,
    y_fa,
    integrator='no_rge'   # The running below 4 GeV is negligible
)
```

```python
chi2_BKll.citation_report('09_results/chi2_BKll_report')
```
