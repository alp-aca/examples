## 6.2 Measurements

In this example, we access the measurement of $B^+ \to K^+ a (\to \mu^+\mu^-)$ at LHCb. For the case of an ALP with $m_a = 1.2\,\mathrm{GeV}$ and $c\tau_a = 100\,\mathrm{cm}$ we obtain the bound and the probability of the ALP decaying with a displaced vertex detected by LHCb.

```python
from alpaca.experimental_data import get_measurements

exp_BKmumu = get_measurements('B+ -> K+ mu mu')
lhcb_BKmumu = exp_BKmumu['LHCb']

central = lhcb_BKmumu.get_central(ma=1.2, ctau=100.0)
sigma_l = lhcb_BKmumu.get_sigma_left(ma=1.2, ctau=100.0)
sigma_r = lhcb_BKmumu.get_sigma_right(ma=1.2, ctau=100.0)
sign = lhcb_BKmumu.decay_type
prob = lhcb_BKmumu.decay_probability(ma=1.2, ctau=100.0)

print(f"Bound = {central} + {sigma_r} - {sigma_l}")
print(f"{sign} decay with probability {prob}")
```

## 6.3 Sectors

Sectors are the tool to group some experimental measurements together in the analysis. The sectors pre-defined in ALPaca are:

```python
for k in alpaca.sectors.default_sectors.keys():
    print(k)
```

`all` is an special sector that includes all the measurements included in the ALPaca database.

We can also define new sectors. For example, this custom sector includes all the Belle II measurements available in ALPaca:

```python
sector_BelleII = alpaca.sectors.Sector(
    name = 'Belle II',
    tex = r'$\mathrm{Belle II}$ bounds',
    observables = [
        'B+ -> K+ e e',
        'B0 -> K*0 e e',
        ('e e -> gamma gamma gamma', 10.58**2),
        'delta_mB0',
        'tau -> e a',
        'tau -> mu a',
    ],
    obs_measurements = {
        'B+ -> K+ mu mu': ['Belle II'],
        'B+ -> K+ a': ['Belle II'],
        'tau -> e e e e': ['Belle II'],
        'tau -> e mu mu mu': ['Belle II'],
        'tau -> mu mu mu': ['Belle II'],
        'tau -> mu e e': ['Belle II'],
    },
    description = 'Belle II measurements implemented in ALPaca.'
)
```

```python
sector_BelleII
```

Another way of creating new sectors is by combining together existing ones:

```python
sector_invisible = alpaca.sectors.combine_sectors(
    [
        alpaca.sectors.default_sectors['BKinv'],
        alpaca.sectors.default_sectors['Kpiinv']
    ],
    name = 'MesonsInv',
    tex = r'$M_1 \to M_2 + \mathrm{inv}$',
    description = 'B -> K inv and K -> pi inv observables.'
)
```

```python
sector_invisible
```

Sectors can be saved in `YAML` files:

```python
sector_BelleII.save('06_results/sector_BelleII.yaml')
sector_invisible.save('06_results/sector_MesonsInv.yaml')
```

which can be later loaded into ALPaca

```python
alpaca.sectors.Sector.load('06_results/sector_BelleII.yaml')
```

It is also possible to load all the sectors stored in a folder, keeping them in a dict:

```python
alpaca.sectors.initialize_sectors('06_results')
```
